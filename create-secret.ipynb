# Criação do secret scope `conn-db`
Notebook isolado do restante do projeto (nunca importado por
`jobs/` ou `src/`), conforme pedido no enunciado. O valor da URI é
digitado no widget abaixo em tempo de execução — **nunca fica
hardcoded no código-fonte nem é versionado no Git**.

```python
dbutils.widgets.text("mongodb_uri", "", "MongoDB URI (não cole aqui um valor que será commitado)")
```

```python
import requests

# Fetch current workspace URL and authentication token automatically
instance = dbutils.notebook.entry_point.getDbutils().notebook().getContext().apiUrl().get()
token = dbutils.notebook.entry_point.getDbutils().notebook().getContext().apiToken().get()

headers = {"Authorization": f"Bearer {token}"}
payload = {
    "scope": "conn-db",
    "initial_manage_principal": "users"  # Options: "users" or "creator"
}

try:
    response = requests.post(f"{instance}/api/2.0/secrets/scopes/create", headers=headers, json=payload)
except Exception as e:
    pass #print(f"Error: {e}")

if response.status_code in (200, 400):
    print("Secret scope 'conn-db' created successfully.")
```

```python
import requests

instance = dbutils.notebook.entry_point.getDbutils().notebook().getContext().apiUrl().get()
token = dbutils.notebook.entry_point.getDbutils().notebook().getContext().apiToken().get()

mongodb_uri = dbutils.widgets.get("mongodb_uri")
if not mongodb_uri:
    raise ValueError(
        "Preencha o widget 'mongodb_uri' com a URI real antes de rodar esta célula "
        "(Run > Edit widget, ou clique no campo no topo do notebook)."
    )

headers = {"Authorization": f"Bearer {token}"}
payload = {
    "scope": "conn-db",
    "key": "cnn-mongodb-sampleflix",
    "string_value": mongodb_uri,
}

response = requests.post(f"{instance}/api/2.0/secrets/put", headers=headers, json=payload)

if response.status_code == 200:
    print("Secret criada/atualizada com sucesso!")
else:
    print(f"Erro ({response.status_code}): {response.text}")
```

```python
dbutils.widgets.remove("mongodb_uri")  # evita que o valor digitado fique retido no estado do notebook
```
